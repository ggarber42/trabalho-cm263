#### renomeando e categorizando

```python
df = df.reset_index(drop=True)
```

```python
rename_columns = {
    'IDSOLICITACAO': 'idsolicitacao',
    'IDPACIENTE': 'idpaciente',
}

df = df.rename(columns=rename_columns)
```

```python
df['situacao_solicitacao'] = df['SITUACAOSOLICITACAO'].astype('category')
df['corregulador'] = df['CORREGULADOR'].astype('category')
df['especialidade_mae'] = df['ESPECIALIDADEMAE'].astype('category')
df['cid'] = df['CID'].astype('category')
df['cnes_solicitante'] = df['CNES_SOLICITANTE'].astype('category')
df['cnesexecutante'] = df['CNESEXECUTANTE'].astype('category')
df['situacao_agenda'] = df['SITUACAOAGENDA'].astype('category')
df['possui_unidade_indicada'] = df['POSSUIUNIDADEINDICADA'].astype('category')
```

```python
drop_columns_2 = [
    'SITUACAOSOLICITACAO',
    'CORREGULADOR',
    'ESPECIALIDADEMAE',
    'CNES_SOLICITANTE',
    'CNESEXECUTANTE',
    'SITUACAOAGENDA',
    'POSSUIUNIDADEINDICADA',
    'CID',
    'UNIDADEEXECUTANTE'
]

df = df.drop(columns=drop_columns_2, errors='ignore')
```

```python
df['cid_ajustada'] = df['cid'].astype(str).str[:3]
```

```python
DATE_FORMAT = '%Y-%m-%d %H:%M:%S.%f'

date_cols = ['DATA_SOLICITACAO', 'DATAHORA_CONSULTA']

for col in date_cols:
    df[col] = pd.to_datetime(
        df[col],
        format=DATE_FORMAT,
        errors='coerce'
    )
```

```python
df['data_solicitacao'] = df['DATA_SOLICITACAO']
df['datahora_consulta'] = df['DATAHORA_CONSULTA']
```

```python
drop_columns_3 = [
    'DATA_SOLICITACAO',
    'DATAHORA_CONSULTA',
]

df = df.drop(columns=drop_columns_3, errors='ignore')
```

```python

df['idade'] = df['IDADE'].astype(float)
df['dias_fila_espera'] = df['DIASFILAESPERA'].astype(float)

df = df.drop(columns=['DIASFILAESPERA', 'IDADE'], errors='ignore')
```

```python
n_dup = df.duplicated().sum()
print(f"Linhas duplicadas: {n_dup:,} ({n_dup / len(df) * 100:.2f}%)")
```
